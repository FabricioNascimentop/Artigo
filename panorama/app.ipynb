```python
import pandas as pd

# 1. Carregar os dados 
df_cadunico = pd.read_csv('../cadunico/panoramaCadunico/sexo_cadunico.csv')
df_ibge = pd.read_csv('../ibge/panorama1020/sexo_ibge.csv')

# 2. Criar uma coluna de ano unificada para o cruzamento
# Como o IBGE tem 2010 e o CadÚnico tem 2013, substituímos 2010 por 2013 no IBGE
df_ibge['Ano_merge'] = df_ibge['Ano'].replace({2010: 2013})
df_cadunico['Ano_merge'] = df_cadunico['Ano']

# 3. Renomear as colunas de acordo com o resultado final desejado
df_cadunico = df_cadunico.rename(columns={
    'Homens': 'homens_cadunico', 
    'Mulheres': 'mulheres_cadunico'
})

df_ibge = df_ibge.rename(columns={
    'Homens': 'homens_geral', 
    'Mulheres': 'mulheres_geral'
})

# 4. Fazer o merge das duas bases usando o Codigo e a nova coluna de Ano
df_merge = pd.merge(
    df_cadunico, 
    df_ibge[['Codigo', 'Ano_merge', 'homens_geral', 'mulheres_geral']], 
    on=['Codigo', 'Ano_merge'], 
    how='left'
)

# 5. Selecionar a ordem final das colunas (mantendo a coluna 'Ano' original do CadÚnico)
colunas_finais = [
    'Codigo', 'Municipio', 'homens_cadunico', 'homens_geral', 
    'mulheres_cadunico', 'mulheres_geral', 'Ano'
]

dfsexo_final = df_merge[colunas_finais]
dfsexo_final['homens_cadunico_%'] = round((dfsexo_final['homens_cadunico'] / dfsexo_final['homens_geral']) * 100,2)
dfsexo_final['mulheres_cadunico_%'] = round((dfsexo_final['mulheres_cadunico'] / dfsexo_final['mulheres_geral']) * 100,2)

dfsexo_final
```

```python
dfsexo_final.to_csv('sexo_geral.csv',index=False)
```

```python
import pandas as pd

# =========================
# 1. Ler os arquivos
# =========================

df_cad = pd.read_csv('../cadunico/panoramaCadunico/faixa_etaria_cadunico.csv')
df_ibge = pd.read_csv('../ibge/panorama1020/faixa_etaria_ibge.csv')


# =========================
# 2. Ajustar os anos do CadÚnico
# =========================

df_cad = df_cad.copy()

df_cad['Ano'] = df_cad['Ano'].replace({
    2013: 2010,
    2022: 2022
})


# =========================
# 3. Colunas das faixas
# =========================

faixas = [
    '0-3',
    '4-6',
    '7-15',
    '16-17',
    '18-24',
    '25-34',
    '35-39',
    '40-44',
    '45-49',
    '50-54',
    '55-59',
    '60-64',
    '64+',
    'Total'
]


# =========================
# 4. Renomear as colunas
# =========================

df_cad = df_cad.rename(columns={
    faixa: f'{faixa}_cadunico'
    for faixa in faixas
})

df_ibge = df_ibge.rename(columns={
    faixa: f'{faixa}_geral'
    for faixa in faixas
})


# =========================
# 5. Fazer o merge
# =========================

df = df_cad.merge(
    df_ibge,
    on=['Codigo', 'Ano'],
    how='outer'
)


# =========================
# 6. Resultado
# =========================

df.to_csv('faixa_etaria_geral.csv',index=False)
```

```python
import pandas as pd

# ==========================
# 1. Ler os arquivos
# ==========================

df_cad = pd.read_csv('../cadunico/panoramaCadunico/etnia_cadunico.csv')
df_ibge = pd.read_csv('../ibge/panorama1020/etnia_ibge.csv')


# ==========================
# 2. Ajustar o código do IBGE
# ==========================

df_ibge['Codigo'] = df_ibge['Codigo'].astype(int)
df_cad['Codigo'] = df_cad['Codigo'].astype(int)


# ==========================
# 3. Fazer correspondência dos anos
# CadÚnico:
# 2013 -> 2010
# 2022 -> 2022
# ==========================

df_cad['Ano_merge'] = df_cad['Ano'].replace({
    2013: 2010,
    2022: 2022
})


# ==========================
# 4. Renomear colunas
# ==========================

df_cad = df_cad.rename(columns={
    'Branca': 'Branca_cadunico',
    'Preta': 'Preta_cadunico',
    'Amarela': 'Amarela_cadunico',
    'Parda': 'Parda_cadunico',
    'Indigena': 'Indigena_cadunico',
    'NoData': 'NoData_cadunico'
})

df_ibge = df_ibge.rename(columns={
    'Branca': 'Branca_geral',
    'Preta': 'Preta_geral',
    'Amarela': 'Amarela_geral',
    'Parda': 'Parda_geral',
    'Indígena': 'Indigena_geral',
    'Total': 'Total_geral'
})


# ==========================
# 5. Merge pelo Código + Ano
# ==========================

df_final = pd.merge(
    df_cad,
    df_ibge,
    left_on=['Codigo', 'Ano_merge'],
    right_on=['Codigo', 'Ano'],
    how='outer',
    suffixes=('_cad', '_ibge')
)


# ==========================
# 6. Organizar Ano
# ==========================

df_final['Ano'] = df_final['Ano_merge']

df_final = df_final.drop(
    columns=['Ano_merge', 'Ano_cad', 'Ano_ibge'],
    errors='ignore'
)


# ==========================
# 7. Remover duplicidades
# ==========================

df_final = df_final.drop_duplicates(
    subset=['Codigo', 'Ano']
)


# ==========================
# 8. Ver resultado
# ==========================

df_final.to_csv('etnia_geral.csv',index=False)
```

```python
import pandas as pd

# ============================================================
# 1. Ler os arquivos
# ============================================================

df_cad = pd.read_csv('../cadunico/panoramaCadunico/urbano_rural_cadunico.csv')
df_ibge = pd.read_csv('../ibge/panorama1020/urbano_rural_ibge.csv')


# ============================================================
# 2. Garantir que Codigo seja numérico
# ============================================================

df_ibge['Codigo'] = df_ibge['Codigo'].astype(int)
df_cad['Codigo'] = df_cad['Codigo'].astype(int)


# ============================================================
# 3. CADÚNICO
#    Há vários registros por município/ano.
#    Usaremos dezembro como referência.
# ============================================================

df_cad = (
    df_cad
    .groupby(['Codigo', 'Ano'], as_index=False)
    .tail(1)
    .copy()
)


# ============================================================
# 4. Ajustar os anos
#
# CadÚnico   IBGE
# 2013   ->   2010
# 2022   ->   2022
# ============================================================

df_cad['Ano_merge'] = df_cad['Ano'].replace({
    2013: 2010,
    2022: 2022
})


# ============================================================
# 5. Renomear as variáveis
# ============================================================

df_cad = df_cad.rename(columns={
    'Rural': 'Rural_cadunico',
    'Urbano': 'Urbano_cadunico'
})

df_ibge = df_ibge.rename(columns={
    'Rural': 'Rural_geral',
    'Urbana': 'Urbana_geral',
    'Total': 'Total_geral'
})


# ============================================================
# 6. Remover colunas desnecessárias do IBGE
# ============================================================

df_ibge = df_ibge.drop(
    columns=['Município', 'Municipio', 'RI'],
    errors='ignore'
)


# ============================================================
# 7. Fazer o MERGE
#
# Codigo é a identificação do município
# Ano_merge faz a correspondência temporal
# ============================================================

df_final = pd.merge(
    df_cad,
    df_ibge,
    left_on=['Codigo', 'Ano_merge'],
    right_on=['Codigo', 'Ano'],
    how='outer'
)


# ============================================================
# 8. Organizar o ano definitivo
# ============================================================

df_final['Ano'] = df_final['Ano_merge']

df_final = df_final.drop(
    columns=['Ano_merge'],
    errors='ignore'
)


# ============================================================
# 9. Manter apenas as colunas relevantes
# ============================================================

df_final = df_final[
    [
        'Codigo',
        'Municipio',
        'RI',
        'Ano',

        'Urbano_cadunico',
        'Rural_cadunico',

        'Urbana_geral',
        'Rural_geral',
        'Total_geral'
    ]
]


# ============================================================
# 10. Ordenar
# ============================================================

df_final = df_final.sort_values(
    ['Codigo', 'Ano']
).reset_index(drop=True)


df_final
```

```python
df_cad = pd.read_csv('../cadunico/panoramaCadunico/urbano_rural_cadunico.csv')
df_ibge = pd.read_csv('../ibge/panorama1020/urbano_rural_ibge.csv')
```

```python
import pandas as pd

# =========================
# 1. Criar ano de referência
# =========================

df_cad['Ano_merge'] = df_cad['Ano'].replace({
    2013: 2010,
    2022: 2022
})

# No IBGE, o próprio Ano será usado como referência
df_ibge['Ano_merge'] = df_ibge['Ano']


# =========================
# 2. Padronizar nomes
# =========================

df_ibge = df_ibge.rename(columns={
    'Município': 'Municipio',
    'Urbana': 'Urbano'
})


# =========================
# 3. Renomear variáveis
#    para não duplicar colunas
# =========================

df_cad = df_cad.rename(columns={
    'Rural': 'Rural_Cadunico',
    'Urbano': 'Urbano_Cadunico'
})

df_ibge = df_ibge.rename(columns={
    'Rural': 'Rural_IBGE',
    'Urbano': 'Urbano_IBGE'
})


# =========================
# 4. Merge
# =========================

df = pd.merge(
    df_cad,
    df_ibge[
        [
            'Codigo',
            'Ano_merge',
            'Municipio',
            'RI',
            'Rural_IBGE',
            'Urbano_IBGE'
        ]
    ],
    on=['Codigo', 'Ano_merge'],
    how='outer',
    suffixes=('_Cadunico', '_IBGE')
)


# =========================
# 5. Unificar Município e RI
# =========================

df['Municipio'] = (
    df['Municipio_Cadunico']
    .fillna(df['Municipio_IBGE'])
)

df['RI'] = (
    df['RI_Cadunico']
    .fillna(df['RI_IBGE'])
)


# =========================
# 6. Remover colunas
#    duplicadas
# =========================

df = df.drop(columns=[
    'Municipio_Cadunico',
    'Municipio_IBGE',
    'RI_Cadunico',
    'RI_IBGE'
])


# =========================
# 7. Organizar resultado
# =========================

df = df[
    [
        'Codigo',
        'Municipio',
        'RI',
        'Ano_merge',
        'Ano',
        'Rural_Cadunico',
        'Urbano_Cadunico',
        'Rural_IBGE',
        'Urbano_IBGE'
    ]
]

df = df[df['Ano_merge'].isin([2010, 2022])]

df.to_csv('../panorama/urbano_rural_geral.csv',index=False)
```
