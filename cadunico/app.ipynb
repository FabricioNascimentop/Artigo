```python
import pandas as pd

dfEtnia = pd.read_csv('dados_prontos/etnia_cadunico_tratado.csv')

dfEtniaCenso = dfEtnia[(dfEtnia['Ano'] == 2013) | (dfEtnia['Ano'] == 2022)]
dfEtniaCenso.rename(columns={'Brancos':'Branca','Pretos':'Preta','Amarelos':'Amarela','Pardos':'Parda','Indigenas':'Indigena'},inplace=True)
dfEtniaCenso
```

```python
dfEtniaCenso.to_csv('panoramaCadunico/etnia_cadunico.csv',index=False)
```
