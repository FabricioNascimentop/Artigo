```python
import pandas as pd

dfEtnia = pd.read_csv('dados_prontos/etnia_cadunico_tratado.csv')


dfEtnia = dfEtnia[dfEtnia['RI'] == 'Guamá']
dfEtniaCenso = dfEtnia[(dfEtnia['Ano'] == 2012) | (dfEtnia['Ano'] == 2022)]
dfEtniaCenso.head(6)
```

```python
import pandas as pd
import plotly.express as px

# Categorias que serão visualizadas
categorias = [
    "Brancos",
    "Pretos",
    "Amarelos",
    "Pardos",
    "Indigenas"
]

# Criar dataframe em formato longo
df_long = dfEtniaCenso.melt(
    id_vars=["Codigo", "Municipio", "Ano", "RI"],
    value_vars=categorias,
    var_name="Raça/Cor",
    value_name="Quantidade"
)

fig = px.bar(
    df_long,
    x="Municipio",
    y="Quantidade",
    color="Raça/Cor",
    facet_row="Ano",
    hover_data={
        "Municipio": True,
        "RI": True,
        "Ano": True,
        "Quantidade": ":,.0f"
    },
    title="Composição racial dos municípios do Pará",
    labels={
        "Municipio": "Município",
        "Quantidade": "Quantidade de pessoas",
        "Raça/Cor": "Raça/Cor"
    },
    barmode="group"
)

fig.update_layout(
    height=900,
    xaxis_tickangle=-45,
    legend_title="Raça/Cor",
    hovermode="x unified"
)

fig.show()
```
